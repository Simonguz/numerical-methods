### Задание 1

Для варианта 7 задана функция $f(x)=\sin(3x)$ и точка $\xi=0{,}7$.

Необходимо найти значение производной функции в точке $x=\xi$ с точностью $10^{-3}$ и $10^{-6}$.

Для вычисления производной можно использовать центральную разностную формулу: $f'(x)\approx \frac{f(x+h)-f(x-h)}{2h}$.

Точное значение производной запрещено использовать как критерий остановки. Для оценки погрешности используется правило Рунге: $R = \frac{|D_{h/2}-D_h|}{2^p-1}$, где для центральной разностной формулы $p=2$.

Вычисления продолжаются до тех пор, пока выполняется условие $R < \varepsilon$.

---

```python
import math

xi = 0.7

def f(x):
    return math.sin(3 * x)

def df(x):
    return 3 * math.cos(3 * x)

def central(x, h):
    return (f(x + h) - f(x - h)) / (2 * h)
```

```python
def runge(eps, h0=0.1):
    p = 2 # порядок точности ! 

    h = h0
    D_h = central(xi, h)

    while True:
        h2 = h / 2
        D_h2 = central(xi, h2)

        runge_err = abs(D_h2 - D_h) / (2**p - 1)

        if runge_err < eps:
            return h2, D_h2, runge_err

        h = h2
        D_h = D_h2

print(f"Пункт 1 \t f(x) = sin(3x) \t xi = {xi}")

for eps in [1e-3, 1e-6]:
    h, D_h2, runge_err = runge(eps)
    print(f"\n Точность eps = {eps} \t h = {h} \t f'(xi) ≈ {D_h2} \t погрешность Рунге = {runge_err} ")
```

### Задание 2

Для функции $f(x)=\sin(3x)$ и точки $\xi=0{,}7$ необходимо сравнить погрешности двух формул численного дифференцирования.

Формула первого порядка точности: $f'(x)\approx \frac{f(x+h)-f(x)}{h}$.

Формула второго порядка точности: $f'(x)\approx \frac{f(x+h)-f(x-h)}{2h}$.

Сравнение выполняется для последовательности убывающих шагов: $h=\frac12,\frac14,\frac18,\frac1{16},\ldots$

Точное значение производной используется только для проверки погрешности: $f'(x)=3\cos(3x)$.

Следовательно, для варианта 7: $f'(0{,}7)=3\cos(2{,}1)$.

Погрешность каждой формулы вычисляется по формуле: $\varepsilon = |f'(\xi)-D_h|$, где $D_h$ — приближённое значение производной при шаге $h$.

Необходимо определить, с какой скоростью убывают погрешности при уменьшении шага. Для формулы первого порядка ожидается $\varepsilon = O(h)$, а для центральной формулы ожидается $\varepsilon = O(h^2)$.

---

```python
def forward(x, h):
    return (f(x + h) - f(x)) / h
```

```python
exact = df(xi)
print(f"Пункт 2. Точное значение: f'(xi) = {exact:.12f}\n")

print(
    f"{'h':>12} | "
    f"{'правая формула':>18} | {'ошибка':>14} | {'коэф.':>8} | "
    f"{'центральная формула':>22} | {'ошибка':>14} | {'коэф.':>8}"
)

print("-" * 112)

prev_err_f = None
prev_err_c = None

for k in range(1, 12):
    h = 1 / (2 ** k)

    D_f = forward(xi, h)
    D_c = central(xi, h)

    err_f = abs(D_f - exact)
    err_c = abs(D_c - exact)

    if prev_err_f is None:
        coef_f = "-"
    else:
        coef_f = f"{prev_err_f / err_f:.4f}"

    if prev_err_c is None:
        coef_c = "-"
    else:
        coef_c = f"{prev_err_c / err_c:.4f}"

    print(
        f"{h:12.8f} | "
        f"{D_f:18.10f} | {err_f:14.6e} | {coef_f:>8} | "
        f"{D_c:22.10f} | {err_c:14.6e} | {coef_c:>8}"
    )

    prev_err_f = err_f
    prev_err_c = err_c
```

### Задание 3

Необходимо исследовать неустойчивость численного дифференцирования для функции $f(x)=\sin(3x)$ в точке $\xi=0{,}7$.

Используется правая разностная формула: $f'(x)\approx \frac{f(x+h)-f(x)}{h}$.

Шаг выбирается из последовательности $h=\frac12,\frac14,\frac18,\frac1{16},\ldots$ или в общем виде $h=2^{-k}$.

Погрешность вычисляется по формуле: $\varepsilon = \left| f'(\xi)-\frac{f(\xi+h)-f(\xi)}{h} \right|$.

Необходимо проверить, будет ли погрешность монотонно убывать при уменьшении шага $h$.

Теоретически при достаточно больших шагах ошибка должна убывать как $O(h)$. Однако при слишком малых значениях $h$ возникает ошибка округления, так как значения $f(x+h)$ и $f(x)$ становятся очень близкими. Из-за этого при их вычитании теряются значащие цифры, и погрешность начинает расти.

---

```python
print("Пункт 3 \nh -> 0\n")

print(
    f"{'k':>4} | "
    f"{'h = 2^-k':>14} | "
    f"{'производная':>18} | "
    f"{'ошибка':>12} | "
    f"{'коэф.':>10}"
)

print("-" * 70)

min_error = float("inf")
min_k = float("inf")
min_h = float("inf")

prev_err = None

for k in range(1, 61):
    h = 2 ** (-k)
    D = forward(xi, h)
    err = abs(D - exact)

    if prev_err is None:
        coef = "-"
    else:
        coef = f"{prev_err / err:.4f}" if err != 0 else "inf"

    if err < min_error:
        min_error = err
        min_k = k
        min_h = h

    print(
        f"{k:4d} | "
        f"{h:14.4e} | "
        f"{D:18.10f} | "
        f"{err:12.4e} | "
        f"{coef:>10}"
    )

    prev_err = err

print(
    f"\nМинимальная ошибка:"
    f"\nk = {min_k}"
    f"\nh = {min_h:.16e}"
    f"\nmin error = {min_error:.16e}"
)
```
