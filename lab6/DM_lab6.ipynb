
# Задание 1

Матрицей Гильберта называется матрица

$$
H_n = \left(\frac{1}{i+j-1}\right), \quad i,j=1,2,\ldots,n.
$$

Для $n = 2,4,6,\ldots,14$ нужно:

- сформировать матрицу Гильберта $H_n$;
- найти число обусловленности $\operatorname{cond}(H_n)$ с помощью `np.linalg.cond`;
- вычислить вектор $f = H_nx$, где $x=(1,1,\ldots,1)^T$;
- решить систему $H_nz=f$ с помощью `np.linalg.solve`;
- вывести покомпонентную разность $z-x$.


```python

def hilbert_matrix(n):
    # Создает матрицу Гильберта размера n x n.
    H = np.zeros((n, n), dtype=float)
    for i in range(n):
        for j in range(n):
            H[i, j] = 1 / (i + j + 1)
    return H

summary_rows = []
diff_rows = []

for n in range(2, 15, 2):
    H = hilbert_matrix(n)
    x = np.ones(n)
    f = H @ x
    z = np.linalg.solve(H, f)
    diff = z - x
    
    summary_rows.append({
        "n": n,
        "cond(H_n)": np.linalg.cond(H),
        "max |z - x|": np.max(np.abs(diff))
    })
    
    row = {"n": n}
    for i, value in enumerate(diff, start=1):
        row[f"z{i}-x{i}"] = value
    diff_rows.append(row)

summary_df = pd.DataFrame(summary_rows)
diff_df = pd.DataFrame(diff_rows)

print("Числа обусловленности и максимальная ошибка:")
display(summary_df)

print("Покомпонентная разность z - x:")
display(diff_df)
```

Чем больше $n$, тем больше число обусловленности $\operatorname{cond}(H_n)$.

Если матрица хорошо обусловлена, то разность $z-x$ должна быть близка к нулю. Для матриц Гильберта при больших $n$ число обусловленности становится очень большим, поэтому даже при решении системы с известным точным ответом появляются заметные вычислительные ошибки.



# Задание 2

Для варианта 7 дана система:

$$
\begin{cases}
8x+y+z=10,\\
x+6y+z=8,\\
x+y+7z=9.
\end{cases}
$$

В матричном виде:

$$
Ax=f,
$$

где

$$
A =
\begin{pmatrix}
8 & 1 & 1 \\
1 & 6 & 1 \\
1 & 1 & 7
\end{pmatrix},
\quad
f =
\begin{pmatrix}
10 \\
8 \\
9
\end{pmatrix}.
$$


```python

A = np.array([
    [8, 1, 1],
    [1, 6, 1],
    [1, 1, 7]
], dtype=float)

f = np.array([10, 8, 9], dtype=float)

print("Матрица A:")
display(pd.DataFrame(A, columns=["x", "y", "z"]))

print("Вектор f:")
display(pd.DataFrame(f, columns=["f"]))
```


## Решение системы напрямую

Сначала решим систему стандартной функцией:

$$
x = A^{-1}f.
$$

В коде используем `np.linalg.solve(A, f)`.


```python

x_direct = np.linalg.solve(A, f)
residual_direct = f - A @ x_direct

print("Решение, полученное np.linalg.solve:")
direct_df = pd.DataFrame({
    "переменная": ["x", "y", "z"],
    "значение": x_direct
})
display(direct_df)

print("Невязка r = f - A x:")
residual_direct_df = pd.DataFrame({
    "компонента": ["r1", "r2", "r3"],
    "значение": residual_direct
})
display(residual_direct_df)
```


## Метод Зейделя

Метод Зейделя строит последовательность приближений $x^{(k)}$.

Для каждой компоненты используется формула:

$$
x_i^{(k+1)} = \frac{1}{a_{ii}}\left(f_i - \sum_{j<i}a_{ij}x_j^{(k+1)} - \sum_{j>i}a_{ij}x_j^{(k)}\right).
$$

В качестве начального приближения возьмем

$$
x^{(0)}=(0,0,0)^T.
$$

Невязку будем считать по формуле:

$$
r^{(k)} = f - Ax^{(k)}.
$$

Для контроля сходимости будем использовать норму:

$$
\|r^{(k)}\|_{\infty}=\max_i |r_i^{(k)}|.
$$


```python

def gauss_seidel(A, f, x0=None, eps=1e-10, max_iter=100):
    # Метод Зейделя для системы A x = f.
    A = np.array(A, dtype=float)
    f = np.array(f, dtype=float)
    n = len(f)
    
    if x0 is None:
        x = np.zeros(n)
    else:
        x = np.array(x0, dtype=float).copy()
    
    history = []
    
    for k in range(1, max_iter + 1):
        x_old = x.copy()
        
        for i in range(n):
            left_sum = np.dot(A[i, :i], x[:i])
            right_sum = np.dot(A[i, i+1:], x_old[i+1:])
            x[i] = (f[i] - left_sum - right_sum) / A[i, i]
        
        residual = f - A @ x
        residual_inf = np.linalg.norm(residual, ord=np.inf)
        residual_2 = np.linalg.norm(residual, ord=2)
        
        history.append({
            "iteration": k,
            "x": x[0],
            "y": x[1],
            "z": x[2],
            "||r||_inf": residual_inf,
            "||r||_2": residual_2
        })
        
        if residual_inf < eps:
            break
    
    return x, pd.DataFrame(history)

x_seidel, seidel_history = gauss_seidel(A, f, eps=1e-10)

print("Решение методом Зейделя:")
display(pd.DataFrame({
    "переменная": ["x", "y", "z"],
    "значение": x_seidel
}))

print("Невязки на каждой итерации метода Зейделя:")
display(seidel_history)

print(f"Количество итераций: {len(seidel_history)}")
```


## Метод верхней релаксации

Метод верхней релаксации отличается от метода Зейделя тем, что новое значение берется с весом $\omega$:

$$
x_i^{(k+1)} = (1-\omega)x_i^{(k)} + \frac{\omega}{a_{ii}}\left(f_i - \sum_{j<i}a_{ij}x_j^{(k+1)} - \sum_{j>i}a_{ij}x_j^{(k)}\right).
$$

При $\omega=1$ метод верхней релаксации совпадает с методом Зейделя.

Проверим несколько значений $\omega$ из диапазона $[1;2]$ и выберем то, при котором сходимость самая быстрая.


```python

def sor(A, f, omega, x0=None, eps=1e-10, max_iter=500):
    # Метод верхней релаксации для системы A x = f.
    A = np.array(A, dtype=float)
    f = np.array(f, dtype=float)
    n = len(f)
    
    if x0 is None:
        x = np.zeros(n)
    else:
        x = np.array(x0, dtype=float).copy()
    
    history = []
    
    for k in range(1, max_iter + 1):
        x_old = x.copy()
        
        for i in range(n):
            left_sum = np.dot(A[i, :i], x[:i])
            right_sum = np.dot(A[i, i+1:], x_old[i+1:])
            x_gs = (f[i] - left_sum - right_sum) / A[i, i]
            x[i] = (1 - omega) * x_old[i] + omega * x_gs
        
        residual = f - A @ x
        residual_inf = np.linalg.norm(residual, ord=np.inf)
        residual_2 = np.linalg.norm(residual, ord=2)
        
        history.append({
            "omega": omega,
            "iteration": k,
            "x": x[0],
            "y": x[1],
            "z": x[2],
            "||r||_inf": residual_inf,
            "||r||_2": residual_2
        })
        
        if residual_inf < eps:
            break
    
    return x, pd.DataFrame(history)
```


## Подбор оптимального значения $\omega$

Переберем значения $\omega$ от $1$ до $2$ с шагом $0.01$.

Оптимальным будем считать то значение $\omega$, при котором метод достиг точности быстрее всего, то есть за минимальное количество итераций.


```python

omega_grid = np.round(np.arange(1.00, 2.01, 0.01), 2)
summary_sor = []

for omega in omega_grid:
    x_sor, history_sor = sor(A, f, omega, eps=1e-10, max_iter=500)
    final_residual = history_sor["||r||_inf"].iloc[-1]
    converged = final_residual < 1e-10
    
    summary_sor.append({
        "omega": omega,
        "iterations": len(history_sor),
        "final ||r||_inf": final_residual,
        "converged": converged,
        "x": x_sor[0],
        "y": x_sor[1],
        "z": x_sor[2]
    })

summary_sor_df = pd.DataFrame(summary_sor)
converged_sor_df = summary_sor_df[summary_sor_df["converged"]].copy()
best_sor = converged_sor_df.sort_values(["iterations", "omega"]).iloc[0]

print("Первые 15 лучших значений omega по числу итераций:")
display(converged_sor_df.sort_values(["iterations", "omega"]).head(15))

print("Лучшее найденное значение omega:")
display(pd.DataFrame([best_sor]))
```


## Невязки метода верхней релаксации для нескольких $\omega$

Выведем невязки на каждой итерации для нескольких значений:

$$
\omega \in \{1.00, 1.01, 1.10, 1.20, 1.50\}.
$$

Значение $\omega=1.00$ оставлено для сравнения, потому что при $\omega=1$ метод верхней релаксации совпадает с методом Зейделя.


```python

omegas_to_show = [1.00, 1.01, 1.10, 1.20, 1.50]
all_histories = []

for omega in omegas_to_show:
    x_sor, history_sor = sor(A, f, omega, eps=1e-10, max_iter=500)
    all_histories.append(history_sor)

sor_histories_df = pd.concat(all_histories, ignore_index=True)

print("Невязки на каждой итерации для выбранных omega:")
display(sor_histories_df)
```

1. Для варианта 7 точное решение системы:

$$
x=1, \quad y=1, \quad z=1.
$$

2. Метод Зейделя сходится к этому решению за несколько итераций.

3. Метод верхней релаксации при $\omega=1$ совпадает с методом Зейделя.

4. По перебору значений $\omega$ с шагом $0.01$ оптимальное значение получилось около:

$$
\omega \approx 1.01.
$$

5. При больших значениях $\omega$ сходимость становится медленнее, поэтому для данной системы слишком сильная верхняя релаксация не нужна.
