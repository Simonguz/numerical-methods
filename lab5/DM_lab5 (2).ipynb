## Пункт 1

Задана функция

$$
f(x) = x^3
$$

на отрезке

$$
[0, 1].
$$

Очевидно, что определённый интеграл от функции $f(x)$ на данном отрезке равен

$$
\int_0^1 x^3 \, dx = \frac{1}{4}.
$$

Необходимо написать программу, вычисляющую значение интеграла по двум формулам:

- формуле трапеций;
- формуле Симпсона.

Также необходимо:

1. Найти максимальную теоретическую ошибку для каждого метода.
2. Найти реальную погрешность, то есть абсолютное значение разности между численным и аналитическим решением:

$$
\Delta = |I_{\text{числ}} - I_{\text{точн}}|.
$$

3. Объяснить, почему при вычислении интеграла по формуле Симпсона для функции $f(x) = x^3$ ошибка равна нулю.
4. Провести численный эксперимент для функций:

$$
f_2(x) = x^2,
$$

$$
f_1(x) = \frac{x}{2}
$$

на отрезке $[0, 1]$ и сравнить полученные погрешности.

```python
import math


def trapezoid(f, a, b, n):
    """
    Составная формула трапеций.
    n — количество отрезков разбиения.
    """
    h = (b - a) / n

    s = 0.5 * f(a) + 0.5 * f(b)

    for i in range(1, n):
        x = a + i * h
        s += f(x)

    return h * s


def simpson(f, a, b, n):
    """
    Составная формула Симпсона.
    n обязательно должно быть чётным.
    """
    if n % 2 != 0:
        raise ValueError("Для формулы Симпсона n должно быть чётным")

    h = (b - a) / n

    s = f(a) + f(b)

    for i in range(1, n):
        x = a + i * h

        if i % 2 == 0:
            s += 2 * f(x)
        else:
            s += 4 * f(x)

    return h * s / 3
```

```python
def trapezoid_error_bound(a, b, n, max_second_derivative):
    """
    Оценка ошибки формулы трапеций:

    |R_T| <= ((b - a) / 12) * h^2 * max|f''(x)|
    """
    h = (b - a) / n
    return ((b - a) / 12) * h**2 * max_second_derivative


def simpson_error_bound(a, b, n, max_fourth_derivative):
    """
    Оценка ошибки формулы Симпсона:

    |R_S| <= ((b - a) / 180) * h^4 * max|f''''(x)|
    """
    h = (b - a) / n
    return ((b - a) / 180) * h**4 * max_fourth_derivative
```

```python
print("Пункт 1")
print("Интегрирование функции f(x) = x^3 на отрезке [0, 1]\n")

a = 0
b = 1
n = 10

f = lambda x: x**3
exact = 1 / 4

trap_value = trapezoid(f, a, b, n)
simp_value = simpson(f, a, b, n)

trap_error = abs(trap_value - exact)
simp_error = abs(simp_value - exact)

# Для f(x) = x^3:
# f''(x) = 6x, максимум на [0,1] равен 6
# f''''(x) = 0
max_f2 = 6
max_f4 = 0

trap_bound = trapezoid_error_bound(a, b, n, max_f2)
simp_bound = simpson_error_bound(a, b, n, max_f4)

print(f"{'Метод':<20} | {'Значение':>18} | {'Реальная ошибка':>18} | {'Теор. ошибка':>18}")
print("-" * 83)

print(f"{'Трапеции':<20} | {trap_value:18.12f} | {trap_error:18.12e} | {trap_bound:18.12e}")
print(f"{'Симпсон':<20} | {simp_value:18.12f} | {simp_error:18.12e} | {simp_bound:18.12e}")

print("\nТочное значение интеграла:")
print(f"I = {exact}")
```

```python
print("Численный эксперимент для функций x^3, x^2, x/2\n")

functions = [
    {
        "name": "x^3",
        "f": lambda x: x**3,
        "exact": 1 / 4,
        "max_f2": 6,
        "max_f4": 0
    },
    {
        "name": "x^2",
        "f": lambda x: x**2,
        "exact": 1 / 3,
        "max_f2": 2,
        "max_f4": 0
    },
    {
        "name": "x/2",
        "f": lambda x: x / 2,
        "exact": 1 / 4,
        "max_f2": 0,
        "max_f4": 0
    }
]

a = 0
b = 1
n = 10

print(
    f"{'Функция':<10} | "
    f"{'Точно':>12} | "
    f"{'Трапеции':>14} | {'Ошибка T':>12} | {'Теор. T':>12} | "
    f"{'Симпсон':>14} | {'Ошибка S':>12} | {'Теор. S':>12}"
)

print("-" * 120)

for item in functions:
    f = item["f"]
    exact = item["exact"]

    trap_value = trapezoid(f, a, b, n)
    simp_value = simpson(f, a, b, n)

    trap_error = abs(trap_value - exact)
    simp_error = abs(simp_value - exact)

    trap_bound = trapezoid_error_bound(a, b, n, item["max_f2"])
    simp_bound = simpson_error_bound(a, b, n, item["max_f4"])

    print(
        f"{item['name']:<10} | "
        f"{exact:12.8f} | "
        f"{trap_value:14.10f} | {trap_error:12.4e} | {trap_bound:12.4e} | "
        f"{simp_value:14.10f} | {simp_error:12.4e} | {simp_bound:12.4e}"
    )
```


## Пункт 2

Используя соотношение

$$
\int_0^1 \frac{1}{1 + x^2} \, dx = \operatorname{arctg}(1),
$$

найти значение числа $\pi$ с точностью

$$
10^{-6}.
$$

Так как

$$
\operatorname{arctg}(1) = \frac{\pi}{4},
$$

то

$$
\pi = 4 \int_0^1 \frac{1}{1 + x^2} \, dx.
$$

В процессе вычислений нельзя использовать встроенную константу $\pi$ для определения величины шага.

Необходимо объяснить, из каких соображений выбирался шаг $h$ для получения указанной точности.

```python
print("Пункт 2")
print("Нахождение числа pi через интеграл\n")

f = lambda x: 1 / (1 + x**2)

a = 0
b = 1

eps = 1e-6

# Для f(x) = 1 / (1 + x^2)
# На [0,1] можно взять оценку max|f''(x)| <= 2
max_f2 = 2

# Ошибка интеграла по формуле трапеций:
# |R| <= ((b-a)^3 / (12 n^2)) * max|f''(x)|
#
# Так как pi = 4I, то ошибка pi будет в 4 раза больше.
#
# Нужно:
# 4 * ((b-a)^3 / (12 n^2)) * max|f''(x)| <= eps

n = math.ceil(
    math.sqrt(
        4 * (b - a)**3 * max_f2 / (12 * eps)
    )
)

h = (b - a) / n

integral_value = trapezoid(f, a, b, n)
pi_value = 4 * integral_value

print(f"eps = {eps}")
print(f"Выбранное количество разбиений n = {n}")
print(f"Шаг h = {h}")
print(f"Полученное значение pi = {pi_value:.12f}")

# math.pi используется только для проверки,
# а не для выбора шага h
real_error = abs(pi_value - math.pi)

print(f"Реальная ошибка = {real_error:.12e}")
```

## Пункт 3

Реализовать предыдущее задание, определяя точность методом Рунге.

При численном вычислении интегралов последовательно с шагами $h$ и $\frac{h}{2}$ можно сократить число арифметических операций.

Заметим, что приближённое значение интеграла $I_{h/2}$ есть сумма, часть слагаемых которой уже участвовала при вычислении $I_h$.

Поэтому можно получить значение $I_{h/2}$, используя ранее найденное значение $I_h$. Это позволяет избежать повторного суммирования части слагаемых.

Для формулы трапеций:

$$
I_h =
h \left(
\frac{1}{2} f_0 + f_1 + \ldots + f_{n-1} + \frac{1}{2} f_n
\right)
$$

При уменьшении шага в два раза:

$$
I_{h/2} =
\frac{h}{2}
\left(
\frac{1}{2} f_0 + f_{1/2} + f_1 + f_{3/2} + \ldots + f_{n-1} + f_{n-1/2} + \frac{1}{2} f_n
\right)
$$

Тогда значение интеграла с шагом $\frac{h}{2}$ можно выразить через $I_h$:

$$
I_{h/2} =
\frac{I_h}{2}
+
\frac{h}{2}
\left(
f_{1/2} + f_{3/2} + \ldots + f_{n - 1/2}
\right)
$$

Для оценки погрешности используется метод Рунге:

$$
R \approx \frac{|I_{h/2} - I_h|}{2^p - 1}
$$

где $p$ — порядок точности метода.

Для формулы трапеций:

$$
p = 2
$$

Следовательно:

$$
R \approx \frac{|I_{h/2} - I_h|}{3}
$$

Так как в предыдущем пункте вычисляется число $\pi$ по формуле

$$
\pi = 4 \int_0^1 \frac{1}{1 + x^2} \, dx,
$$

то оценку ошибки интеграла нужно умножить на $4$:

$$
R_{\pi} \approx 4 \cdot \frac{|I_{h/2} - I_h|}{3}
$$

Вычисления продолжаются до тех пор, пока выполняется условие:

$$
R_{\pi} < 10^{-6}
$$

```python
def trapezoid_refine(f, a, b, old_integral, n):
    """
    Уточнение интеграла по формуле трапеций.

    Был шаг h.
    Теперь делаем шаг h/2.

    Используем старое значение интеграла,
    чтобы не пересчитывать все точки заново.
    """
    h = (b - a) / n

    s_new = 0

    for i in range(n):
        x_middle = a + (i + 0.5) * h
        s_new += f(x_middle)

    new_integral = old_integral / 2 + h * s_new / 2

    return new_integral
```

```python
print("Пункт 3")
print("Нахождение pi с точностью 1e-6 методом Рунге\n")

f = lambda x: 1 / (1 + x**2)

a = 0
b = 1

eps = 1e-6

p = 2

n = 1
I_h = trapezoid(f, a, b, n)

print(
    f"{'№':>4} | "
    f"{'n':>8} | "
    f"{'h':>14} | "
    f"{'pi':>18} | "
    f"{'оценка ошибки':>18}"
)

print("-" * 75)

iteration = 0

while True:
    iteration += 1

    I_h2 = trapezoid_refine(f, a, b, I_h, n)

    # оценка ошибки интеграла
    integral_error_estimate = abs(I_h2 - I_h) / (2**p - 1)

    # так как pi = 4I
    pi_error_estimate = 4 * integral_error_estimate

    n = n * 2
    h = (b - a) / n

    pi_value = 4 * I_h2

    print(
        f"{iteration:4d} | "
        f"{n:8d} | "
        f"{h:14.6e} | "
        f"{pi_value:18.12f} | "
        f"{pi_error_estimate:18.6e}"
    )

    if pi_error_estimate < eps:
        break

    I_h = I_h2

print("\nОтвет:")
print(f"pi ≈ {pi_value:.12f}")
print(f"n = {n}")
print(f"h = {h:.12e}")
print(f"Оценка ошибки по Рунге = {pi_error_estimate:.12e}")

# Только для проверки
print(f"Реальная ошибка = {abs(pi_value - math.pi):.12e}")
```
